```
arr=np.array([[1,2,3,4,5],[4,5,6,7,8]])
print(arr)
print(arr.sum(axis=0))
print(arr.sum(axis=1))
```

```
l=[10,20,30,40]#1d array indexing
arr=np.array(l)
print(arr[3])
print(arr[0])
print(arr[1])
print(arr[-1])
```

```
l2=np.array([   #indexing in array of matrix
    [1,2,3],
    [4,5,6],
    [7,8,9]
])
print(l2)
print(l2[0,0])
print(l2[2,2])
print(l2[1,0])
print(l2[0])
print(l2[1])
print(l2[2])
print(l2[:,0])
print(l2[:,1])
print(l2[:,2])
```

```
#slicing [start:end:step] 1d
l=[1,2,3,4,5]
arr=np.array(l)
print(arr)
print(arr[0:2])
print(arr[0:5])
print(arr[1:5])
print(arr[0:5:2])
print(arr[1:])
print(arr[:4])
```

```
#slicing [row_start:row_end,col_start:col_end] 2d
l=np.array([
    [1,2,3,4],
    [4,5,6,7],
    [8,9,10,11]
])
print(l)
print(l[0:2,0:2])
```
